```python
import pandas as pd
from sklearn.cluster import KMeans

# 1. Load dataset from the data folder
df = pd.read_csv('../data/Mall_Customers.csv')

# 2. Select features (Annual Income & Spending Score)
X = df.iloc[:, [3, 4]].values

# 3. Train K-Means model
kmeans = KMeans(n_clusters=5, init='k-means++', random_state=42)
y_kmeans = kmeans.fit_predict(X)

# 4. Export results to the outputs folder
df['Cluster'] = y_kmeans
df.to_csv('../outputs/customer_clusters.csv', index=False)
print("Saved to task2/outputs/customer_clusters.csv successfully!")
```
